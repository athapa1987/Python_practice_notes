```python
table_of_four = [value * 4 for value in range (1, 11)]
print (table_of_four)
print ('\n above method is comprensive list and the following method define value, loop, add the value then print values outside the list.\n')
multiplication_of_four = []
for values in range (1, 11):
    multiplication_of_four.append(values*4)
print (multiplication_of_four)
```
