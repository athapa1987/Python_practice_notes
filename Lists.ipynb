```python
#slicing the list
next_plan = ['study', 'learn', 'mediate', 'develop', 'calm down', 'deep thinking']
print (next_plan [0]) # just calling the first item from list. 
print (next_plan [0:3]) # slicing and seeing the first item to 3rd item in the list. 
print (next_plan [1:4]) # slicing and seeing the 2nd item to 4th item in the list. 
print (next_plan [:2]) # slicing and seeing the first item 2 item in the list. 
print (next_plan [3:])  # slicing and seeing the items after 3rd item in the list.
print (next_plan [-1])  # slicing and seeing the last item in the list.
```

```python
# Looping through a slice
next_plan = ['study', 'learn', 'mediate', 'develop', 'calm down', 'deep thinking']
print ("Here are the your first three future plans")
for future_plans in next_plan [:3]:
    print (future_plans.upper())
print ('\nThis method could be used to call the desired item from list line by line. \n we can try this with other slicing list of above method.')
```
